# Orientation maps in $(\phi_1,\Delta\phi)$

Companion to [progress_summary.ipynb](progress_summary.ipynb) (Sections 8 and 10). Every figure there that is plotted against
the two initial in-plane moment angles $(\phi_1,\phi_2)$ is redrawn here in
$$\Delta\phi\equiv\phi_2-\phi_1,\qquad \text{wrapped to }[0,2\pi),$$
next to the original. Where it helps, a third panel uses the sum and difference, $(\phi_1+\phi_2,\ \Delta\phi)$.
($\theta$ is not used for this angle: in Bartz & Shaw, arXiv:2212.00876, $\theta$ is the orbital angle of $\hat r$, with
$U\propto3\cos(2\theta-\phi_1-\phi_2)+\cos(\phi_1-\phi_2)$.)

**Wrap choice.** $\Delta\phi\in[0,2\pi)$ puts the antiparallel line $\Delta\phi=\pi$ (the cusp, head-to-head and tail-to-tail)
in the middle of the axis and the parallel line $\Delta\phi=0$ on its edges. Exchanging the spheres, $\Delta\phi\to2\pi-\Delta\phi$,
is then a reflection about the middle. $[-\pi,\pi)$ would split the antiparallel cusp across the top and bottom edges.

**No new runs, no interpolation.** The data are the stored references: the $32\times32$ torus map
(`tests/reference/map_torus32.npz`) and the four 800-scan zoomed maps (`tests/reference/zoom_maps.npz`, the data of
`figures/*_scan800_*.pdf`). Every map samples $\Delta\phi$ on one uniform lattice. The torus has equal $\phi_1$ and $\phi_2$ steps,
and the strips and patches have commensurate steps. So $(\phi_1,\Delta\phi)$ is an **exact index shift**: each $\phi_1$ row is
rolled by an integer number of cells (`analysis/delta_phi.py: delta_grid`, which raises if a shift is not an integer). The
$(\phi_1+\phi_2,\Delta\phi)$ lattice is a checkerboard, not a rectangle, so those panels draw each sample's own cell mapped by
the linear change of variables (diamonds, or parallelograms for the strips). Nothing is resampled, so window edges and the
capture boundary are exactly as sharp as in the originals. Grey means "not sampled".

### The mirror line and the other reference lines

The mirror line is defined in Section 8 of the progress notebook ("How to read the map") as the orientations where $\hat\mu_2$ is
$\hat\mu_1$ reflected across the $y$ axis, $\phi_2=\pi-\phi_1$, and drawn as the dotted lines $\phi_1+\phi_2=\pi$ (and $3\pi$). Section 10 uses the strip
coordinates $\phi_1=s$, $\phi_2=\pi-s+d$, so $d=0$ is the line.

| line | $(\phi_1,\phi_2)$ | $(\phi_1,\Delta\phi)$ | $(\phi_1+\phi_2,\Delta\phi)$ | style |
|---|---|---|---|---|
| mirror | $\phi_1+\phi_2\equiv\pi \pmod{2\pi}$ | $\Delta\phi\equiv\pi-2\phi_1$ (slope $-2$, two branches) | $\phi_1+\phi_2=\pi$ (vertical) | dotted |
| antiparallel | $\phi_2\equiv\phi_1+\pi$ | $\Delta\phi=\pi$ (horizontal) | $\Delta\phi=\pi$ (horizontal) | dashed |
| parallel | $\phi_2\equiv\phi_1$ | $\Delta\phi=0$ (bottom/top edge) | $\Delta\phi=0$ (edge) | dash-dot |

The trough or window band sits just *below* the mirror line ($d<0$). In the new coordinates it is just below the slope $-2$
line at fixed $\phi_1$, and just *left* of $\phi_1+\phi_2=\pi$.

### Symmetries in each coordinate system

| operation | $(\phi_1,\phi_2)$ | $(\phi_1,\Delta\phi)$ | $(\phi_1+\phi_2,\Delta\phi)$ |
|---|---|---|---|
| flip both moments (exact symmetry) | $(\phi_1+\pi,\phi_2+\pi)$ | translation $\phi_1\to\phi_1+\pi$ | identity: flip partners land on the same point |
| exchange the spheres (exact symmetry) | $(\phi_2,\phi_1)$ | $(\phi_1+\Delta\phi,\,-\Delta\phi)$, a shear-reflection | reflection $\Delta\phi\to2\pi-\Delta\phi$ about the antiparallel line |
| $b\to-b$ composed with exchange (not a symmetry of the $b>0$ map) | $(\pi-\phi_2,\pi-\phi_1)$ | — | reflection $\phi_1+\phi_2\to2\pi-(\phi_1+\phi_2)$ about the mirror line |

The last row uses Section 8 of the progress notebook: the run at $-b$ for $(\phi_1,\phi_2)$ equals the run at $+b$ for
$(\pi-\phi_1,\pi-\phi_2)$. Combined with exchange, this means that **reflecting the $b>0$ map across the mirror line gives the
$b<0$ map**.

```python
import os
import sys
from pathlib import Path

ROOT = Path(os.environ["MAGNETORBITS_ROOT"]) if "MAGNETORBITS_ROOT" in os.environ else next(
    p for p in [Path.cwd().resolve(), *Path.cwd().resolve().parents] if (p / "physics").is_dir())
sys.path.insert(0, str(ROOT)); sys.path.insert(0, str(ROOT / "analysis"))

import json
import numpy as np
import matplotlib.pyplot as plt

from physics import central
from tests.published.bartz_shaw_2022 import ENERGY_TILDE as E_PAPER
from delta_phi import (VIEWS, COORDS, TWO_PI, delta_grid, draw_map, line_legend, pi_ticks, sample_angles, wrap)

# Plot style: no titles, no gridlines, large fonts.
plt.rcParams.update({"font.size": 16, "axes.labelsize": 22, "xtick.labelsize": 18, "ytick.labelsize": 18,
                     "legend.fontsize": 15, "axes.grid": False, "legend.frameon": False, "figure.dpi": 80})
FIG = ROOT / "figures"
B_LOCKED = central.capture_impact_parameter(E_PAPER, 2.0)

def save(fig, stem):
    # PDF (committed, like every other figure) and a 200-dpi PNG (enough pixels for the 1024-row dphi panels)
    if os.environ.get("MAGNETORBITS_NO_SAVE"):        # set by tests/test_notebook.py
        return
    for ext, dpi in (("pdf", 200), ("png", 200)):
        fig.savefig(FIG / f"{stem}.{ext}", dpi=dpi)

torus = np.load(ROOT / "tests" / "reference" / "map_torus32.npz")
phi = torus["phi"]; n = len(phi); h = TWO_PI / n
zoom = np.load(ROOT / "tests" / "reference" / "zoom_maps.npz")
LINE = {"strip_anti": "antiparallel", "strip_mirror": "mirror", "patch_tt": "box", "patch_min": "box"}
maps = {k: {"along": zoom[f"{k}__along"], "across": zoom[f"{k}__across"], "b_out": zoom[f"{k}__outer_edge"],
            "n_int": zoom[f"{k}__n_intervals"], "line": LINE[k], **json.loads(str(zoom[f"{k}__meta"]))} for k in LINE}

# Exactness: every sample of every map comes back unchanged from its (phi1, dphi) cell.
for name, (kind, al, ac, v) in {"torus32": ("torus", phi, phi, torus["outer_edge"]),
                                **{k: (m["line"], m["along"], m["across"], m["b_out"]) for k, m in maps.items()}}.items():
    p1, p2 = sample_angles(kind, al, ac)
    grid, x, y, step = delta_grid(v, p1, p2)
    rows = np.round((wrap(p2 - p1) - y[0]) / step).astype(int) % int(round(TWO_PI / step))
    exact = np.array_equal(np.take_along_axis(grid, rows, 1), v, equal_nan=True)
    print(f"{name:13s} {v.shape[0]:3d} x {v.shape[1]:3d} -> {grid.shape[0]:3d} x {grid.shape[1]:4d} (phi1, dphi) cells, "
          f"dphi step pi/{int(round(np.pi / step))}; all samples recovered exactly: {exact}")

# The symmetries the new coordinates rely on, checked on the stored torus map.
I, J = np.meshgrid(np.arange(n), np.arange(n), indexing="ij")
for key in ("outer_edge", "n_intervals", "capture_measure"):
    v = torus[key]
    flip = np.max(np.abs(v - v[(I + n // 2) % n, (J + n // 2) % n]))
    exch = np.max(np.abs(v - v.T))
    print(f"torus {key:15s}: flip max |diff| {flip:.1e}; exchange max |diff| {exch:.1e}")
```

## 1. The $32\times32$ torus map (progress notebook §8)

Left: the original $(\phi_1,\phi_2)$ map with all three reference lines and the reference orientations (HT head-to-tail, HH
head-to-head, P parallel and A antiparallel side-by-side). Middle: the same pixels in $(\phi_1,\Delta\phi)$. Right: the same pixels in
$(\phi_1+\phi_2,\Delta\phi)$, where each pixel becomes a diamond. Each window of the right panel shows every orientation twice,
once for each flip partner, and the two copies have identical values (checked above).

```python
def torus_figure(key, label, cmap, stem, **kw):
    v = torus[key]
    fig, axes = plt.subplots(1, 3, figsize=(22, 7.4), layout="constrained")
    for ax, view in zip(axes, VIEWS):
        im = draw_map(ax, "torus", phi, phi, v, view, cmap=cmap, vmin=np.nanmin(v), vmax=np.nanmax(v), **kw)
    cb = fig.colorbar(im, ax=axes, label=label, shrink=0.92)
    line_legend(fig)
    return fig, cb

fig, cb = torus_figure("outer_edge", r"outer collision edge $b_{\rm out}/R$", "viridis", "torus32_outer_edge_dphi")
cb.ax.axhline(B_LOCKED, color="w", lw=2)            # locked head-to-tail value, as in the original
save(fig, "torus32_outer_edge_dphi"); plt.show()

fig, cb = torus_figure("n_intervals", "number of contact intervals", "magma", "torus32_intervals_dphi")
cb.set_ticks([1, 2]); save(fig, "torus32_intervals_dphi"); plt.show()

fig, cb = torus_figure("capture_measure", r"capture measure $/R$", "viridis", "torus32_measure_dphi")
save(fig, "torus32_measure_dphi"); plt.show()

win = np.argwhere(torus["n_intervals"] == 2)
print("escape-window orientations:", len(win))
print("  phi1 + phi2 (mod 2pi) at the windows, in units of pi/16:", sorted({int((i + j) % n) for i, j in win}))
print("  dphi at the windows, in units of pi/16:                  ", sorted({int((j - i) % n) for i, j in win}))
```

**What changes.**
- In $(\phi_1,\Delta\phi)$ the antiparallel dip is one horizontal row at $\Delta\phi=\pi$, and the parallel line is the bottom
  and top edges. The flip symmetry is now a translation by $\pi$ in $\phi_1$, so the left and right halves of the panel are
  identical. The mirror line becomes two parallel lines of slope $-2$, $\Delta\phi\equiv\pi-2\phi_1$, and the trough and the
  windows run just below them. The mirror features are no easier to see in this panel than in the original.
- In $(\phi_1+\phi_2,\Delta\phi)$ both families are straight and axis-aligned. Antiparallel is horizontal, and the mirror line is
  vertical at $\phi_1+\phi_2=\pi$. **All 32 escape-window orientations fall in the single column $\phi_1+\phi_2=15\pi/16$**, at
  every sampled $\Delta\phi$. The trough's position depends on the sum only. Exchange symmetry is the up-down reflection about
  $\Delta\phi=\pi$, and the flip symmetry is already quotiented out, so this window holds the whole map once.
- The map is visibly *not* left-right symmetric about the mirror line. That is expected: the reflection
  $\phi_1+\phi_2\to2\pi-(\phi_1+\phi_2)$ turns the $b>0$ map into the $b<0$ map (table above). So the "lopsided trough" is the $b>0$
  half of a pair. For $b<0$ the windows sit at $\phi_1+\phi_2=17\pi/16$, just to the right of the line.

### 1D cuts (progress notebook §8, cell 32)

Top row: the original cuts at $\phi_1=2,4,6\,\pi/16$, $\pm6$ cells across the antiparallel line (left) and the mirror line (right).
Bottom left: the matching $(\phi_1,\Delta\phi)$ version. Both original cuts are columns at fixed $\phi_1$, so here each is the whole
column, plotted against $\Delta\phi$. The antiparallel crossing is at $\Delta\phi=\pi$ for every curve (grey dashed). The mirror
crossing $\Delta\phi=\pi-2\phi_1$ moves with $\phi_1$ (dotted, in each curve's color), and the parallel line is at the edges.
Bottom right (optional sum/difference version): cuts at fixed $\Delta\phi$ plotted against $\phi_1+\phi_2-\pi$, the distance
from the mirror line. These use odd multiples of $\pi/16$, because only those rows of the torus lattice pass through the
window column. Filled symbols are the stored $b>0$ edge. Open symbols are the same data reflected across the mirror line, which is
the outer edge at $b<0$.

```python
edge = torus["outer_edge"]
k = np.arange(-6, 7)
fig, axes = plt.subplots(2, 2, figsize=(18, 11), layout="constrained")
(ax1, ax2), (ax3, ax4) = axes
for i0, color in [(2, "C0"), (4, "C1"), (6, "C2")]:
    lab = fr"$\phi_1={i0}\pi/16$"
    ax1.plot(k * h, [edge[i0, (i0 + n // 2 + d) % n] for d in k], "o-", color=color, ms=5, label=lab)
    ax2.plot(k * h, [edge[i0, (n // 2 - i0 + d) % n] for d in k], "o-", color=color, ms=5, label=lab)
    col = np.array([edge[i0, (i0 + m) % n] for m in range(n)])          # dphi = m h, m = 0 .. n-1
    ax3.plot(np.arange(n) * h, col, "o-", color=color, ms=5, label=lab)
    ax3.axvline(wrap(np.pi - 2 * i0 * h), color=color, ls=":", lw=2)
for m, color in [(13, "C3"), (9, "C4"), (5, "C5")]:
    sig = np.array([(2 * i + m) % n for i in range(n // 2)])            # phi1 + phi2 = sig h; one flip partner each
    vals = np.array([edge[i, (i + m) % n] for i in range(n // 2)])
    order = np.argsort(wrap(sig * h - np.pi, -np.pi))
    x = wrap(sig * h - np.pi, -np.pi)[order]
    ax4.plot(x, vals[order], "o-", color=color, ms=6, label=fr"$\Delta\phi={m}\pi/16$, $b>0$")
    ax4.plot(-x, vals[order], "o--", color=color, ms=6, mfc="none", lw=1, label=fr"$\Delta\phi={m}\pi/16$, $b<0$")
for ax in (ax1, ax2, ax4):
    ax.axvline(0, color="0.6", lw=1, ls="--" if ax is ax1 else ":")
ax3.axvline(np.pi, color="0.6", ls="--", lw=1.5)
for ax in axes.flat:
    ax.axhline(B_LOCKED, color="0.3", ls="-", lw=0.8)
    ax.set_ylabel(r"$b_{\rm out}/R$")
ax1.set_xlabel(r"$\phi_2-\phi_1-\pi$   (from antiparallel)")
ax2.set_xlabel(r"$\phi_1+\phi_2-\pi$   (from the mirror line)")
ax3.set_xlabel(r"$\Delta\phi=\phi_2-\phi_1$"); ax3.set_xlim(-h / 2, TWO_PI - h / 2)
ax4.set_xlabel(r"$\phi_1+\phi_2-\pi$   (from the mirror line)"); ax4.set_xlim(-np.pi / 3, np.pi / 3); ax4.set_ylim(4.35, None)
for ax in axes.flat:
    pi_ticks(ax, "x", max_ticks=7)
ax1.legend(fontsize=14, loc="lower left"); ax4.legend(fontsize=12, ncol=3, loc="lower center")
ax2.text(-0.65 * np.pi / 2, B_LOCKED + 0.03, "locked head-to-tail", fontsize=13, color="0.3")
save(fig, "torus32_cuts_dphi"); plt.show()
```

The $\Delta\phi$ column (bottom left) shows that the antiparallel dip sits at the same place for every $\phi_1$. The trough
just below the mirror crossing moves with $\phi_1$, and near HH ($\phi_1=2\pi/16$, blue) the two features nearly coincide. In
the sum coordinate (bottom right), the trough is at $\phi_1+\phi_2-\pi=-\pi/16$ for every $\Delta\phi$, and the $b<0$ curves are its
mirror image at $+\pi/16$. On the mirror line itself ($\phi_1+\phi_2=\pi$) the $b>0$ and $b<0$ edges coincide.

## 2. The antiparallel strip (progress notebook §10a)

The strip coordinates are $\phi_1=s$, $\phi_2=s+\pi+d$, so $d=\Delta\phi-\pi$ exactly, and $(\phi_1,\Delta\phi)$ is the original strip
transposed with its across-axis shifted by $\pi$. That is the natural frame for this line. The sum/difference panel spreads each
$s$ row over $\phi_1+\phi_2=2s+\pi+d$. The cusp stays a horizontal line, and the places where the mirror band crosses it
($s\approx0$ and $s\approx\pi$) both land near $\phi_1+\phi_2=\pi$.

```python
def strip_figure(key, quantity, label, cmap, stem, views=VIEWS, figsize=(22, 7.4)):
    m = maps[key]; v = m[quantity]
    fig, axes = plt.subplots(1, len(views), figsize=figsize, layout="constrained")
    for ax, view in zip(axes, views):
        im = draw_map(ax, m["line"], m["along"], m["across"], v, view, cmap=cmap, vmin=np.nanmin(v), vmax=np.nanmax(v))
        ax.set_facecolor("0.85")
    cb = fig.colorbar(im, ax=axes, label=label, shrink=0.92)
    line_legend(fig)
    save(fig, stem); plt.show()
    return cb

strip_figure("strip_anti", "b_out", r"$b_{\rm out}/R$", "viridis", "strip_anti_scan800_edge_dphi")
strip_figure("strip_anti", "n_int", "contact intervals", "magma", "strip_anti_scan800_intervals_dphi")
```

```python
m = maps["strip_anti"]
s, d, v = m["along"], m["across"], m["b_out"]
j0 = int(np.argmin(np.abs(d)))
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(22, 6.4), layout="constrained")
powers = []
for k in (15, 31, 47):
    lab = fr"$\phi_1={s[k] / np.pi:.2f}\pi$"
    ax1.plot(d, v[k], lw=1.8, label=lab)                       # original: against d
    ax2.plot(np.pi + d, v[k], lw=1.8, label=lab)               # new: dphi = pi + d
    pos = (d > 0) & (d < 0.02 * np.pi)
    powers.append(np.polyfit(np.log(d[pos]), np.log(v[k, pos] - v[k, j0]), 1)[0])
    ax3.loglog(d[pos], v[k, pos] - v[k, j0], "o", ms=4, mfc="none", label=lab)
dd = np.geomspace(2e-3, 2e-2, 10) * np.pi
pos = (d > 0) & (d < 0.02 * np.pi)
amp = np.median((v[31, pos] - v[31, j0]) / d[pos] ** np.mean(powers))
ax3.loglog(dd, amp * dd ** np.mean(powers), "k--", lw=1.2, label=fr"$\propto|\Delta\phi-\pi|^{{{np.mean(powers):.2f}}}$")
for ax in (ax1, ax2):
    ax.axhline(B_LOCKED, color="0.4", ls=":", lw=1)
    ax.axvline(0 if ax is ax1 else np.pi, color="0.6", ls="--", lw=1.2)
    ax.set_ylabel(r"$b_{\rm out}/R$"); pi_ticks(ax, "x")
ax1.set_xlabel(r"$d$   ($\phi_2=\phi_1+\pi+d$)"); ax2.set_xlabel(r"$\Delta\phi=\phi_2-\phi_1$")
ax3.xaxis.set_minor_formatter(plt.NullFormatter()); ax3.yaxis.set_minor_formatter(plt.NullFormatter())
ax3.set_xlabel(r"$\Delta\phi-\pi$  (rad)"); ax3.set_ylabel(r"$b_{\rm out}(\Delta\phi)-b_{\rm out}(\pi)$")
ax1.legend(fontsize=14); ax3.legend(fontsize=14)
save(fig, "strip_anti_scan800_profiles_dphi"); plt.show()
print(f"cusp exponent {np.mean(powers):.2f} (progress notebook: 0.57); identical data, x shifted by pi")
```

## 3. The mirror strip and the window band (progress notebook §10b)

The strip coordinates are $\phi_1=s$, $\phi_2=\pi-s+d$, so $d=\phi_1+\phi_2-\pi$ and $\Delta\phi=\pi-2s+d$. In $(\phi_1,\Delta\phi)$ (middle) the
strip becomes a slanted band of slope $-2$ that wraps once around $\Delta\phi$, and the window band runs diagonally. This
coordinate does not straighten the mirror line. In $(\phi_1+\phi_2,\Delta\phi)$ (right) the strip is a vertical band around the
mirror line, and it covers all of $\Delta\phi$ once. That is the original strip picture with its along-axis converted from
$s=\phi_1$ to $\Delta\phi$.

```python
strip_figure("strip_mirror", "b_out", r"$b_{\rm out}/R$", "viridis", "strip_mirror_scan800_edge_dphi", figsize=(22, 8.2))
strip_figure("strip_mirror", "n_int", "contact intervals (lower bound)", "magma", "strip_mirror_scan800_intervals_dphi",
             figsize=(22, 8.2))

m = maps["strip_mirror"]
s, d = m["along"], m["across"]
print(" phi1/pi   dphi on the line/pi   windows at (phi1+phi2-pi)/pi      b_out just below the band")
for k in range(0, 64, 8):
    w = m["n_int"][k] >= 2
    j = int(np.argmin(m["b_out"][k]))
    print(f"  {s[k] / np.pi:.3f}      {wrap(np.pi - 2 * s[k]) / np.pi:.3f}              "
          f"[{d[w].min() / np.pi:+.4f}, {d[w].max() / np.pi:+.4f}]             {m['b_out'][k, j]:.3f} at {d[j] / np.pi:+.4f}")
# Band edges against the distance from the antiparallel line, measured where the band's lower edge is
# (each strip row is a slanted line in (phi1 + phi2, dphi), so dphi is taken at the edge itself).
lower = np.array([d[m["n_int"][k] >= 2].min() for k in range(len(s))])
upper = np.array([d[m["n_int"][k] >= 2].max() for k in range(len(s))])
dist = np.abs(wrap(np.pi - 2 * s + lower - np.pi, -np.pi))          # |dphi - pi| at the lower edge
for cut in (0.3 * np.pi, 0.2 * np.pi, 0.1 * np.pi):
    sel = dist >= cut
    print(f"|dphi - pi| >= {cut / np.pi:.1f} pi ({sel.sum():2d} rows): lower edge {lower[sel].min() / np.pi:+.4f} .. "
          f"{lower[sel].max() / np.pi:+.4f} pi, upper edge {upper[sel].min() / np.pi:+.4f} .. {upper[sel].max() / np.pi:+.4f} pi")
print(f"closest to antiparallel (|dphi - pi| = {dist.min() / np.pi:.3f} pi): lower edge {lower[np.argmin(dist)] / np.pi:+.4f} pi; "
      f"grid step pi/512 = {1 / 512:.4f} pi")
```

## 5. Summary: which symmetries are clearer in the new coordinates

**$(\phi_1,\Delta\phi)$, with $\Delta\phi\in[0,2\pi)$.**
- *The antiparallel and parallel families are horizontal lines* ($\Delta\phi=\pi$, $\Delta\phi=0$). The antiparallel cusp is a
  horizontal valley of nearly constant depth ($b_{\rm out}\approx4.6$–$4.72$ for $0.1\pi<\phi_1<0.9\pi$) that deepens only at
  $\phi_1\to0,\pi$, where the window band crosses it. For the antiparallel strip this is the original frame, with $d=\Delta\phi-\pi$.
- *The flip symmetry is a translation*, $\phi_1\to\phi_1+\pi$, so every full map is $\pi$-periodic in $\phi_1$.
- *The mirror line is not straightened.* It becomes $\Delta\phi\equiv\pi-2\phi_1$, two lines of slope $-2$. The trough, the window
  band and the profiles across it all sit at a $\Delta\phi$ that moves with $\phi_1$, and exchange symmetry becomes a
  shear-reflection that is hard to see by eye. For the mirror-line structure, $(\phi_1,\Delta\phi)$ is no clearer than $(\phi_1,\phi_2)$.

**$(\phi_1+\phi_2,\ \Delta\phi)$ (the optional panel; included because it straightens the mirror line).**
- *Both families are straight and perpendicular.* The mirror line is vertical ($\phi_1+\phi_2=\pi$) and antiparallel is horizontal
  ($\Delta\phi=\pi$). HH and TT both sit at their crossing.
- *The window band depends on the sum only.* On the torus all 32 window orientations sit at $\phi_1+\phi_2=15\pi/16$, a single
  column. On the mirror strip the band covers the same interval, $-0.078\pi\le\phi_1+\phi_2-\pi\le-0.002\pi$, at every
  $\Delta\phi$ with $|\Delta\phi-\pi|\ge0.3\pi$, to the strip's $\pi/512$ grid. Its sharp lower edge is a vertical line there. It
  narrows only near the antiparallel line, to about $-0.055\pi$.
- *Exchange is the reflection $\Delta\phi\to2\pi-\Delta\phi$ about the antiparallel line*, visible as up-down symmetry in every panel
  (exact in the data). *The flip symmetry is built in:* flip partners map to the same point.
- *Left-right about the mirror line is $b\to-b$, not a symmetry.* Reflecting the $b>0$ map across $\phi_1+\phi_2=\pi$ gives the
  $b<0$ map, and on the torus they differ at 94% of points (by up to 0.83 in $b_{\rm out}$). The lopsided trough just left of the line
  ($15\pi/16$) has a $b<0$ partner just right of it ($17\pi/16$). On the line itself, and on $\phi_1+\phi_2=0$, the two signs of $b$
  have the same outer edge. A two-sided quantity, such as the planar cross section
  $\sigma=\text{measure}(\phi_1+\phi_2,\Delta\phi)+\text{measure}(2\pi-\phi_1-\phi_2,\Delta\phi)$, would be symmetric about both axes.

**Caveats.** The interval counts in the zoomed maps are lower bounds inside the window band, and the patch minimum is a lower
bound near the crossing (progress notebook §10). The coordinate change does not alter these limits. The PNGs are written at
200 dpi so the 1024-row $\Delta\phi$ axis of the mirror-strip panels is not downsampled; the inline previews here are smaller.
